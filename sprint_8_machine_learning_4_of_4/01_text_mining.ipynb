### 1. Data Preparation
In this first part, we will explore our dataset and clean it. The dataset we will use contains excerpts from Donald Trump's speeches during his 2016 campaign.

**(a)** Using the `pandas` package, import the data contained in the file `"trump.csv"` into a `DataFrame` named `df` using the following separator : `'|'`. 

**(b)** Display the first 10 rows of the dataset.

```python
import pandas as pd

# (a) Import the data
df = pd.read_csv("trump.csv", sep="|")

# (b) Display the first 10 rows
df.head(10)
```

**Output Description:** The code displays the first 10 rows of the DataFrame, showing the 'Speech' and 'Sentiment' columns.

**(c)** Create a string named `speeches` containing the concatenation of all entries in the `"Speech"` column of `df`. Be sure to insert a space between each line.

```python
# (c) Concatenate all speeches into a single string
speeches = " ".join(df["Speech"].astype(str))
```

**Output Description:** This cell creates a single string variable named `speeches` containing all the text from the 'Speech' column. There is no direct output displayed.

In the text, the audience reactions are indicated by square brackets, `[Applause]` for example.

**(d)** Create a `test` variable containing the string: `Hello World! [Applause] How are you ?`.

**(e)** Using the `re` library, create a function named `remove_brackets` which takes as input a string and replaces the words in square brackets with spaces.

**(f)** Apply this function to `test` to check its effectiveness.

```python
import re

# (d) Create a test string
test = "Hello World! [Applause] How are you ?"

# (e) Define a function to remove text in brackets
def remove_brackets(text):
    return re.sub(r'\[.*?\]', '', text)

# (f) Apply the function to the test string
print(remove_brackets(test))
```

**Output Description:** The output shows the test string with the `[Applause]` part removed, verifying the function works correctly.

**(g)** Apply the `remove_brackets` function to the `speeches` string.

```python
# (g) Apply the function to the main speeches string
speeches = remove_brackets(speeches)
```

**Output Description:** This cell cleans the main `speeches` string by removing all bracketed text. No output is displayed.

**(h)** Using the `nltk` library, create a variable named `stop_words` containing the common stop words of the English language and some additional custom words and characters.

```python
import nltk
from nltk.corpus import stopwords

# (h) Create a set of stopwords
# nltk.download('stopwords') # Uncomment and run this line if you haven't downloaded stopwords yet
stop_words = set(stopwords.words('english')).union({
    "?", "!", ".", ",", ":", ";", "-", "--", "...", "\"", "'", 
    "they've", "they're", "they'll", "i've", "i'm", "i'll", "could"
})
```

**Output Description:** This cell defines a set of words that will be filtered out from the text. No output is displayed.

**(i)** Import the `TweetTokenizer` class from the `nltk.tokenize` package.

**(j)** Instantiate the `TweetTokenizer` class in a `tokenizer` object.

**(k)** Apply the `tokenize` method of `tokenizer` to `speeches` and store the result in a variable named `tokens`.

```python
from nltk.tokenize import TweetTokenizer

# (i, j) Instantiate the tokenizer
tokenizer = TweetTokenizer()

# (k) Tokenize the speeches string
# nltk.download('punkt') # Uncomment and run this line if you haven't downloaded the tokenizer models yet
tokens = tokenizer.tokenize(speeches)
```

**Output Description:** The `speeches` string is broken down into a list of individual words (tokens). No output is displayed.

**(l)** Display the total number of words as well as the number of different words found in these speeches.

```python
# (l) Print total and unique word counts before filtering
print(f"Total words: {len(tokens)}")
print(f"Unique words: {len(set(tokens))}")
```

**Output Description:** The output shows the total count of all words (70180) and the count of unique words (4938) in the speeches before filtering.

**(m)** Remove all `stop_words` from the `tokens` list and display the new total word count.

```python
# (m) Filter out stopwords from the tokens list
tokens = [word for word in tokens if word.lower() not in stop_words]

# Display the new counts
print(f"Total words after filtering: {len(tokens)}")
print(f"Unique words after filtering: {len(set(tokens))}")
```

**Output Description:** The output shows the new total word count (28383) and unique word count (4679) after removing the common stop words.

### 2. Wordcloud Construction
In this section we will build a wordcloud from the `tokens` variable.

**(a)** Import the `WordCloud` class.

**(b)** Instantiate the `WordCloud` class with specific parameters.

**(c)** Display the wordcloud using `matplotlib.pyplot`.

```python
import numpy as np
from PIL import Image
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Load the mask image
mask = np.array(Image.open("trump.jpg"))

# (a, b) Instantiate the WordCloud object
wc = WordCloud(
    background_color="white",
    max_words=1000,
    max_font_size=90,
    collocations=False,
    random_state=42,
    mask=mask
)

# Generate the wordcloud from tokens
wc.generate(" ".join(tokens))

# (c) Display the wordcloud
plt.figure(figsize=(10, 10))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```

**Output Description:** An image of a word cloud is displayed, shaped like the provided `trump.jpg` mask. The most frequent words in the speeches appear largest. The words are in default colors on a white background.

**(d)** Import the `ImageColorGenerator` class.

**(e)** Instantiate an `ImageColorGenerator` object.

**(f)** Use the `recolor` method to apply the image's colors to the wordcloud.

**(g)** Remove the axes from the graph.

**(h)** Display the wordcloud again.

```python
from wordcloud import ImageColorGenerator

# (d, e) Create a color generator from the mask image
img_color = ImageColorGenerator(mask)

# (f) Recolor the wordcloud using the image colors
wc.recolor(color_func=img_color)

# (g, h) Display the final, recolored wordcloud
plt.figure(figsize=(10, 10))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```

**Output Description:** A new word cloud image is displayed. It has the same shape and words as before, but now the colors of the words are taken from the original `trump.jpg` image.

### 3. Sentiment Analysis
In this part, we will try to create a sentiment analysis model on this dataset.

**(a)** Import the `train_test_split` function.

**(b)** Create a `Series` `X` and a `Series` `y`.

**(c)** Split the data into training and testing sets.

```python
from sklearn.model_selection import train_test_split

# (b) Define features (X) and target (y)
X = df["Speech"]
y = df["Sentiment"]

# (c) Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=20)
```

**Output Description:** The data is split into training and testing sets for both features (`X`) and the target variable (`y`). No output is displayed.

**(d)** Import the `CountVectorizer` class.

**(e)** Instantiate the `CountVectorizer` class.

**(f)** Apply the `fit_transform` and `transform` methods on the data.

```python
from sklearn.feature_extraction.text import CountVectorizer

# (d, e) Instantiate the vectorizer
vectorizer = CountVectorizer()

# (f) Vectorize the text data
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

**Output Description:** This cell converts the text data into a numerical format (a matrix of token counts) that machine learning models can understand. No output is displayed.

**(g)** Train a `DecisionTreeClassifier`.

**(h)** Train a `GradientBoostingClassifier`.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import GradientBoostingClassifier

# (g) Train a Decision Tree model
decision_tree = DecisionTreeClassifier()
decision_tree.fit(X_train_vec, y_train)

# (h) Train a Gradient Boosting model
gradient_boosting = GradientBoostingClassifier()
gradient_boosting.fit(X_train_vec, y_train)
```

**Output Description:** The two classification models are trained on the vectorized training data. The output shows the default parameters of the `GradientBoostingClassifier`.

**(i)** Using the accuracy on the test set, compare the performance of the two algorithms.

```python
from sklearn.metrics import accuracy_score

# (i) Predict and calculate accuracy for both models
dt_pred = decision_tree.predict(X_test_vec)
gb_pred = gradient_boosting.predict(X_test_vec)

dt_acc = accuracy_score(y_test, dt_pred)
gb_acc = accuracy_score(y_test, gb_pred)

print(f"Decision Tree Accuracy: {dt_acc:.4f}")
print(f"Gradient Boosting Accuracy: {gb_acc:.4f}")
```

**Output Description:** The accuracy scores for both models are printed. The `GradientBoostingClassifier` performs better than the `DecisionTreeClassifier`.

**(j)** Display the confusion matrices of these two algorithms on the test set. Which sentiment is difficult to detect?

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Calculate confusion matrices
dt_cm = confusion_matrix(y_test, dt_pred)
gb_cm = confusion_matrix(y_test, gb_pred)

# Get class labels
labels = sorted(y_test.unique())

# Plot Decision Tree Confusion Matrix
plt.figure(figsize=(6, 5))
sns.heatmap(dt_cm, annot=True, fmt="d", cmap="Blues", xticklabels=labels, yticklabels=labels)
plt.title("Decision Tree Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

# Plot Gradient Boosting Confusion Matrix
plt.figure(figsize=(6, 5))
sns.heatmap(gb_cm, annot=True, fmt="d", cmap="Greens", xticklabels=labels, yticklabels=labels)
plt.title("Gradient Boosting Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

**Output Description:** Two heatmaps are displayed, representing the confusion matrices for each model.
* **Top-left to bottom-right diagonal:** Correct predictions.
* **Off-diagonal:** Incorrect predictions.

By comparing the off-diagonal values, we can see how each sentiment is being misclassified. Both models have more trouble correctly identifying the **Negative** sentiment compared to the Positive one. Therefore, **'Negative' is the more difficult sentiment to detect**.
